Task 6: Read File Safely (Error Handling Inside File Handling Only)

```python
import os

file_name = input("Enter the file name to open")
valid_extensions = ('.txt', 'ipynb')
has_valid_ext = file_name.lower().endswith(valid_extensions)

if has_valid_ext == True:
    if os.path.exists(file_name):
        with open(file_name, "r") as file:
            content = file.read()
            print(content)
    else:
        print("File not found")
else:
    print("File does not contain an acceptable extension")

```
