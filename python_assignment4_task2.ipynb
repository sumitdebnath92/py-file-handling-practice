Task 2: Read File in Diffrent Ways

```python
with open("sales_data.txt", "r") as file:
    read_file = file.read()
    print(read_file)

    file.seek(0)
    read_line = file.readline()
    print(read_line)

    file.seek(0)
    read_all_line = file.readlines()
    print(read_all_line)

    integer_list = []
    for each_line in read_all_line:
        for each_item in each_line.strip().split(','):
            if each_item.split():
                integer_list.append(int(each_item))
       
    print(integer_list)

  
```
