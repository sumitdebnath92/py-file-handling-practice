Task 1: Write Sales Records to a File

```python
sales = [1200, 450, 980, 3000]

sales_data_file = open("sales_data.txt","w")
for sales_amount in sales:
    sales_data_file.writelines(str(sales_amount)+"\n")
sales_data_file.close()

with open("sales_data.txt", "r") as file:
    content = file.read()
    print(content)

with open("sales_data.txt","a") as sales_data_file: 
    sales_data_file.write("\n")
    for sales_amount in sales:
        if(sales.index(sales_amount)<len(sales)-1):
            sales_data_file.write(f"{str(sales_amount)},")
        else:
            sales_data_file.write(f"{str(sales_amount)}")

with open("sales_data.txt", "r") as file:
    content = file.read()
    print(content)    
```
