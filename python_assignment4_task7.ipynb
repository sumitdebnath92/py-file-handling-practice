Task 7: Mini Project - Export Discount Prices

```python
prices = {"Mouse":500, "Keyboard":800, "Monitor":7000, "Pendrive":400, "Camera":5000} 

discount = float(input("Enter discount percentage (e.g., 10 or 0.10):"))

discount_percentage = discount / 100 if discount > 1 else discount

with open("discount_report.txt","w") as discount_report_file:
    for key in prices:
        discount_report_file.writelines(f"{key}|{prices.get(key)}|{int(prices.get(key)) - int(prices.get(key)) * discount_percentage}\n")

print(f"{'Product List':<20} - {'Product Price':<20} - Product Discounted Price")

total_item = 0
total_discount = 0.0

with open("discount_report.txt", "r") as read_file:
    for line in read_file:
        if line.strip():
            product, price, discounted_price = line.strip().split("|")
            total_item += 1
            total_discount += float(discounted_price)
            print(f"{product.strip():<20} - {price.strip():<20} - {discounted_price.strip()}")
print()
print("Summary:")
print(f"Total Items: {total_item}")
print(f"Average Discounted Price: {total_discount/total_item}")



```
