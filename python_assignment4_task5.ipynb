Task 5: Create Product Info File (User Input)

```python
product_details = []
for i in range(3):
    product_name = input("Enter Product Name: ").strip().lower()
    product_price = input(f"Enter {product_name}'s price: ").strip().lower()
    product_dictionary = {"product":product_name,"price":product_price}
    product_details.append(product_dictionary)

print(product_details)

product_details_file = open("product_details_data.txt","w")
for products in product_details:
    product_details_file.writelines(f"{products.get("product")}|{products.get("price")}\n")
product_details_file.close()

print(f"{'Product List':<20} - Product Price")

with open("product_details_data.txt", "r") as read_file:
    for line in read_file:
        if line.strip():
            product, price = line.strip().split("|")
            print(f"{product.strip():<20} - {price.strip()}")

```
