Task 3: Append New Sales

```python
new_list = [5000,2500,1700]
ends_with_comma = False;
with open("sales_data.txt", "r") as file:
    content = file.read()
    if content.rstrip().endswith(","):
        ends_with_comma = True;
    
    print(content)

with open("sales_data.txt","a") as sales_data_file: 
    if(ends_with_comma==False):
        sales_data_file.write(",")
    for sales_amount in new_list:
        if(new_list.index(sales_amount)<len(new_list)-1):
            sales_data_file.write(f"{str(sales_amount)},")
        else:
            sales_data_file.write(f"{str(sales_amount)}")

with open("sales_data.txt", "r") as file:
    content = file.read()
    print(content)
```
