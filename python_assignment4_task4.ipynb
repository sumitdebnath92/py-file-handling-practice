Task 4: Generate Summary Report from File

```python
with open("sales_data.txt", "r") as file:
 
    file.seek(0)
    read_all_line = file.readlines()
    print(read_all_line)

    sales_data_int_list = []
    for each_line in read_all_line:
        for each_item in each_line.strip().split(','):
            if each_item.split():
                sales_data_int_list.append(int(each_item))
       
    print(sales_data_int_list)

    total_sales = sum(sales_data_int_list)
    print(f"Total Sales: {total_sales}")

    highest_sales = max(sales_data_int_list)
    print(f"Heighest Sales: {highest_sales}")

    lowest_sales = min(sales_data_int_list)
    print(f"Lowest Sales: {lowest_sales}")

    average_sales = total_sales/len(sales_data_int_list)
    print(f"Average Sales: {average_sales}")

```
